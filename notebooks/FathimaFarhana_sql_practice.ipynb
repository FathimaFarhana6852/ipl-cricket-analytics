```python
query('''
SELECT LOWER(TRIM(bowler_type)) AS folded,
       COUNT(DISTINCT bowler_type)AS spellings
FROM deliveries
WHERE bowler_type IS NOT NULL
GROUP BY folded
HAVING COUNT(DISTINCT bowler_type) > 1;
''')
```
