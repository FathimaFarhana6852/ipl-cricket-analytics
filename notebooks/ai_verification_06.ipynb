## AI Assisted Verification

process of checking answers , calculations,code or findings suggested by AI tools against reliable evidence such as our actual database.

#### Basic AI verification Process:

1. As a specific or checkable question.
2. Record the AI's answer.
3. Write and run your own sql query.
4. Compare both results.
5. Record agreement or disagreement and explain the verdict.

```python
q("""
SELECT bowler,
       SUM(total_runs
           - COALESCE(bye_runs, 0)
           - COALESCE(leg_bye_runs, 0)
           - COALESCE(penalty_runs, 0)
       ) AS runs_conceded,

       SUM(CASE WHEN is_wide_ball = 0
                AND is_no_ball = 0
                THEN 1 ELSE 0
           END) AS legal_balls,

       ROUND(SUM(
               total_runs
               - COALESCE(bye_runs, 0)
               - COALESCE(leg_bye_runs, 0)
               - COALESCE(penalty_runs, 0)) * 6.0 /
           NULLIF(
               SUM(CASE WHEN is_wide_ball = 0
                        AND is_no_ball = 0
                        THEN 1 ELSE 0
                   END), 0
           ), 2
       ) AS economy_rate

FROM v_deliveries_category_clean
WHERE bowler = 'A Kumble'
  AND is_super_over = 0
GROUP BY bowler;
""")
```

```python
q("""
SELECT
    bowler,

    SUM(
        total_runs
        - COALESCE(bye_runs, 0)
        - COALESCE(leg_bye_runs, 0)
        - COALESCE(penalty_runs, 0)
    ) AS runs_conceded,

    SUM(
        CASE
            WHEN is_wide_ball = 0
             AND is_no_ball = 0
            THEN 1
            ELSE 0
        END
    ) AS legal_balls,

    ROUND(
        SUM(
            total_runs
            - COALESCE(bye_runs, 0)
            - COALESCE(leg_bye_runs, 0)
            - COALESCE(penalty_runs, 0)
        ) * 6.0
        /
        NULLIF(
            SUM(
                CASE
                    WHEN is_wide_ball = 0
                     AND is_no_ball = 0
                    THEN 1
                    ELSE 0
                END
            ), 0
        ),
        2
    ) AS economy_rate

FROM v_deliveries_category_clean

WHERE bowler = 'A Kumble'
  AND is_super_over = 0

GROUP BY bowler;
""")
```

```python
from scipy import stats

toss_test = q("""
SELECT
    COUNT(*) AS n,
    SUM(
        CASE
            WHEN toss_winner = match_winner THEN 1
            ELSE 0
        END
    ) AS k
FROM matches
WHERE result = 'win'
  AND toss_winner IS NOT NULL
  AND match_winner IS NOT NULL;
""")

n = toss_test["n"].iloc[0]
k = toss_test["k"].iloc[0]

result = stats.binomtest(
    k=k,
    n=n,
    p=0.5,
    alternative="two-sided"
)

print("P-value:", round(result.pvalue, 5))
```

```python
from scipy import stats

toss_decision_table = q("""
SELECT
    CASE
        WHEN toss_decision = 'bat' THEN 'bat'
        WHEN toss_decision = 'field' THEN 'field'
        ELSE 'unknown'
    END AS toss_decision,

    SUM(CASE
        WHEN toss_winner = match_winner THEN 1
        ELSE 0
    END) AS toss_winner_won,

    SUM(CASE
        WHEN toss_winner != match_winner THEN 1
        ELSE 0
    END) AS toss_winner_didnot_win

FROM matches_clean
WHERE result = 'win'
  AND toss_decision IN ('bat', 'field')
GROUP BY toss_decision
ORDER BY toss_decision;
""")

table = toss_decision_table[
    ["toss_winner_won", "toss_winner_didnot_win"]
].values

chi2, p, dof, expected = stats.chi2_contingency(table)

print(round(p, 4))
```

```python
from scipy import stats

toss_decision_table = q("""
SELECT
    CASE
        WHEN toss_decision = 'bat' THEN 'bat'
        WHEN toss_decision = 'field' THEN 'field'
    END AS toss_decision,

    SUM(CASE
        WHEN toss_winner = match_winner THEN 1
        ELSE 0
    END) AS toss_winner_won,

    SUM(CASE
        WHEN toss_winner != match_winner THEN 1
        ELSE 0
    END) AS toss_winner_didnot_win

FROM matches_clean

WHERE result = 'win'
  AND toss_decision IN ('bat', 'field')
  AND toss_winner IS NOT NULL
  AND match_winner IS NOT NULL

GROUP BY toss_decision;
""")

table = toss_decision_table[
    ['toss_winner_won', 'toss_winner_didnot_win']
].values

chi2, p, dof, expected = stats.chi2_contingency(table)

print(round(p, 4))
```
