```python
con.execute("DROP TABLE IF EXISTS matches_clean;")
con.commit()
```

```python
con.executescript("""
CREATE TABLE matches_clean AS
WITH ranked_venues AS(
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY venue
            ORDER BY venue_id
        )AS rn
    FROM venues
),  
dedup_venues AS (
    SELECT *
    FROM ranked_venues
    WHERE rn = 1
)

SELECT
   m.*,

   COALESCE(
        v.city,m.city,
        'UNKNOWN'
    ) AS city_clean,
    CAST(
        SUBSTR(m.season,1,4)
        AS INTEGER)
        AS season_year

FROM v_matches_clean m

LEFT JOIN dedup_venues v
    ON m.venue=v.venue;
""")
con.commit()
```

***ast.literal_eval***=it is a built in python function that safely converts a string containing a basic python data structure directly into its actual python object form.

**batsman strike rate**=it is the average number of run a batters scores per 100 legal balls they faced.

**SR=Batter runs/legal balls faced*100**

high strike rate = the batsmans scores runs very quickly

**bowling strike rate**=it is the average number of balls a bowler needs to bowl to take a wicket

*low strike rate* =the bowler takes wickets quickly

**economy rate**= it is the average number of runs a bowler concedes(give away) for every over they bowl

**qualification threshold**= it means setting a minimum limit of matches,runs,or overs before including a player in a leader board

economy leaderboard=min 500 legal balls

```python
fielder_counts =(
    fielders_exploded['fielders']
    .dropna()
    .value_counts()
)

fielder_counts.head(10)
```

```python
batting=(
    d.groupby('batter')
    .agg(
        runs=('batter_runs','sum')
    )
    .reset_index())
```

```python
batting.sort_values('runs',ascending=False).head()
```

```python
legal_balls=(
    d[
        (d['is_wide_ball']==0)&
        (d['is_no_ball']==0)
    ].groupby('batter').size().reset_index(name='legal_balls')
)
legal_balls.head()
```

```python
batting=batting.merge(
    legal_balls,
    on='batter',
    how='left'
)
batting.head()
```

```python
batting['strike_rate']=(
    batting['runs']/batting['legal_balls']
)*100
batting.sort_values('runs',ascending=False).head()
```

```python
non_bowler_wickets=[
    "run out",
    "retired hurt",
    "retired out",
    "obstructing the field"
]
```

```python
bowling_wickets=(
    bowler_wickets_rows
    .groupby('bowler')
    .size()
    .reset_index(name="wickets")
    .sort_values('wickets',ascending=False)
)
bowling_wickets.head()
```

**bye_runs**= runs score when the ball passes the batter without touching their bat or body and the batters run to exchange ends.

**leg_bye_runs**=runs score when the ball touches the batters body or protecting bats and the batters run to exchange ends.

**penalty_runs**=runs awarded directly by the umpire as a punishment to one team for breaking cricket rules or misbehaving

```python
d['bowler_runs']=(
    d['total_runs']
    -d['bye_runs']
    -d['leg_bye_runs']
    -d['penalty_runs']
)
```

```python
bowler_legal_balls=(
    d[(d['is_wide_ball']==0)&
      (d['is_no_ball']==0)
      ]
).groupby('bowler').size().reset_index(name='legal_balls')


bowler_legal_balls.head(10)
```

```python
bowler_runs=(
    d.groupby('bowler')['bowler_runs']
    .sum()
    .reset_index(name='runs_conceded')
)

bowler_runs.head()
```

```python
bowler=bowler_legal_balls.merge(
    bowler_runs,
    on='bowler',
    how='left'
)
bowler.head()
```

```python
bowler['economy_rate']=(
    bowler['runs_conceded'] /
    (bowler['legal_balls']/6))
```

```python
qualified_bowler=bowler[
    bowler['legal_balls']>=500
]
qualified_bowler.sort_values(
    "economy_rate",
    ascending=True
).head()
```
